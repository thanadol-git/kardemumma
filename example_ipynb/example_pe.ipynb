# ProteomEdge — example usage

Functions from `kardemumma.proteomedge`:

| Function | What it does |
|---|---|
| `extract_lot_number` | Parse a lot number from a URL or plain string |
| `fetch_qreps_table` | Fetch the qRePS table as a DataFrame (no file saved) |
| `load_qRePs` | Fetch + save a dated CSV, return `(df, filename)` |
| `fetch_fasta` | Fetch the lot FASTA file as a DataFrame (no file saved) |
| `save_fasta` | Fetch + save a dated `.fasta` file, return `(df, filename)` |

```python
import kardemumma as kdm
```

## 1. Extract lot number

```python
# From a plain lot number string
print(kdm.extract_lot_number("23002"))

# From a full ProteomEdge URL
print(kdm.extract_lot_number("https://proteomedge.com/lotdata/23002/"))
```

## 2. Fetch qRePS table (in-memory, no file written)

```python
# Pass either a lot number or a full URL
qreps_df = kdm.fetch_qreps_table("23002")
print(qreps_df.shape)
qreps_df.head()
```

## 3. Load qRePS and save to CSV

```python
# Fetches the table and writes a dated CSV: YYYYMMDD_<lot>_qRePs.csv
qreps_df, out_file = kdm.load_qRePs("23002")
print("Saved to:", out_file)
qreps_df.head()
```

## 4. Fetch FASTA (in-memory, no file written)

Scrapes the lot page, follows the `.fasta` download link, and parses
the content into a tidy DataFrame with columns `id`, `description`, `sequence`.

```python
fasta_df = kdm.fetch_fasta("23002")
print(f"{len(fasta_df)} entries")
fasta_df.head()
```

```python
# Inspect sequence lengths
fasta_df["seq_len"] = fasta_df["sequence"].str.len()
fasta_df[["id", "description", "seq_len"]].sort_values("seq_len", ascending=False)
```

## 5. Save FASTA locally

```python
# Saves to YYYYMMDD_<lot>.fasta by default
fasta_df, fasta_file = kdm.save_fasta("23002")
print("Saved to:", fasta_file)
fasta_df.head()
```

```python
# Or specify your own filename
fasta_df, fasta_file = kdm.save_fasta("23002", out_file="lot_23002.fasta")
print("Saved to:", fasta_file)
```
