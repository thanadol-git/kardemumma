# ProteomEdge qRePS table demo

This notebook shows how to use the `skyline_qc` package to download the qRePS concentration table for a given ProteomEdge lot (e.g. `24001`).

You can change the lot number or full URL in the code cell below.

```python
import sys, os

# Show which Python the notebook is using (check it’s the .venv one)
print(sys.executable)

# Make the local package importable while working in this repo
sys.path.insert(0, os.path.abspath("src"))

from skyline_qc.proteomedge import fetch_qreps_table
```

```python
from skyline_qc.proteomedge import fetch_qreps_table

# Set the ProteomEdge lot number or full URL here.
# Examples:
#   lot_or_url = "24001"
#   lot_or_url = "https://proteomedge.com/lotdata/24001/"
lot_or_url = "240031"
```

```python
# Fetch the qRePS table for the given lot and show the first rows
qreps_df = fetch_qreps_table(lot_or_url)

print(f"Loaded {len(qreps_df)} rows and {len(qreps_df.columns)} columns")
qreps_df.head()
```

```python
# (Optional) Save the table to a CSV file in the current folder
output_path = f"qreps_{lot_or_url.strip().strip('/')}.csv"
qreps_df.to_csv(output_path, index=False)
print(f"Saved qRePS table to: {output_path}")
```
